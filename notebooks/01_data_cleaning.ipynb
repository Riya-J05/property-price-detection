```python
col_names = df.columns

for col in col_names:
    null_val = df[col].isna().sum()
    if null_val >= 1:
        print(f'{col}: {null_val} null values')
```

```python
df.describe()
```

```python
# checking for duplicates!
```

```python
df.duplicated().sum()
```

```python
duplicate_mask = df.duplicated(keep=False)

df[duplicate_mask].sort_values(
    by=df.columns.tolist()
)
```

```python
df = df.drop_duplicates()
df.shape
```

```python
# consistency issues in categorical cols!
```

```python
cat_cols = df.select_dtypes(exclude=['number']).columns.tolist()

for col in cat_cols:
    print(df[col].value_counts(), end='\n\n')
```

```python
df['location'] = df['location'].str.strip().str.lower()
df['location'].value_counts()
```

```python
df['parking'] = df['parking'].str.strip().str.lower()
df['parking'].value_counts()
```

```python
df['furnishing'] = df['furnishing'].str.strip().str.replace(' ', '-').str.lower()
df['furnishing'].value_counts()
```

```python
# since we changed some categorical columns to lowercase, let's change to lowercase for the other cols as well!
```

```python
rem_cat_cols = ['property_type', 'has_gym', 'has_security']

for col in rem_cat_cols:
    df[col] = df[col].str.strip().str.lower()
```

```python
# numerical validity checking!
```

```python
df[df['floor'] > df['total_floors']]
```

```python
df.nlargest(10, 'area_sqft')
```

```python
df.nlargest(10, 'monthly_maintenance_inr')
```

```python
# Check negative values in numerical columns
num_cols = df.select_dtypes(include='number').columns

(df[num_cols] < 0).sum()
```
